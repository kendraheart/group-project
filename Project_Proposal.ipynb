# Estimation of Workout Effects From Fast Food Calories

### Kendra Chavez, Elise Frerichs 

[The link to our chosen dataset](https://www.kaggle.com/datasets/samartalwar/fast-food-calorie-shock-food-vs-workout-burn)

There are 8000 instances in our dataset with 23 total attributes for each instance 


**Attribute Descriptions** 
- order_id: A unique alphanumberic order identifier
- fast_food_chain: The chain name where the order was placed
- meal_category: The composition of the category
- item_count: The total number of food or drink items in the meal order
- total_calories: Gross energetic value of the complete order in kilocalories (kcal)
- fat_grams: Total dietary fat content in grams
- saturated_fat_grams: Saturated fatty acids in grams 
- carb_grams: Total carbohydrates in grams
- sugar_grams: The added sugar content in grams
- protein_grams: Total dietary protein in grams
- sodium_mg: Total sodium content in milligrams
- consumer_gender: The gender of the consumer (male or female)
- consumer_age: Chronological age of consumer (18-65 years old)
- consumer_weight_kg: The consumers body mass in kilograms, scaling MET burn formulas
- fitness_level: Cardiovascular conditioning bracket: Sedentary, moderately, active, highly trained
- thermic_effect_calories: Energy consumed by the body to digest and metabolize the macronutrients (TEF)
- net_caloric_surplus: The net caloric burden to be burned via exercise after TEF deduction
- walking_burn_time_min: total minutes of brisk walking required to burn off meal 
- cycling_burn_time_min: total minutes of moderate road cycling required to burn off meal
- running_burn_time_min: total minutes of continuous running required to burn off meal
- hiit_burn_time_min: total minutes of high-intensity calisthenics/HIIT required to burn off meal 
- daily_calorie_allowance_pct: The caloric payload expressed as a percentage of standard 2,000 kcal daily intake 
- workout_shock_tier: Multi-class target classifying cardio penalty: Moderate Burn, High Sweat Penalty, Severe Cardio Deficit, or Extreme, Half-Marathon Toll
